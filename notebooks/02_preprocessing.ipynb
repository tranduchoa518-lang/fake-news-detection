```python
import pandas as pd
import numpy as np
import re
import nltk

from nltk.corpus import stopwords
from nltk.stem import PorterStemmer

nltk.download("stopwords")

print("Đã import thư viện thành công!")
```

```python
from google.colab import files

uploaded = files.upload()

df = pd.read_csv("fake_news_dataset.csv")

df.head()
```

```python
print("Kích thước dữ liệu:", df.shape)

print("\nGiá trị bị thiếu:")
print(df.isnull().sum())

df = df.dropna(subset=["text", "label"])

df["text"] = df["text"].astype(str)

print("\nKích thước sau khi loại bỏ dữ liệu thiếu:", df.shape)
```

```python
stop_words = set(stopwords.words("english"))

stemmer = PorterStemmer()
```

```python
def preprocess_text(text):

    # Chuyển văn bản thành chữ thường
    text = text.lower()

    # Loại bỏ đường dẫn URL
    text = re.sub(r"http\S+|www\S+", "", text)

    # Loại bỏ ký tự đặc biệt và chữ số
    text = re.sub(r"[^a-zA-Z\s]", "", text)

    # Tách từ
    words = text.split()

    # Loại bỏ từ dừng và thực hiện stemming
    words = [
        stemmer.stem(word)
        for word in words
        if word not in stop_words
    ]

    return " ".join(words)
```

```python
df["clean_text"] = df["text"].apply(preprocess_text)

df[["text", "clean_text", "label"]].head()
```

```python
print("Văn bản ban đầu:")
print(df["text"].iloc[0])

print("\nVăn bản sau tiền xử lý:")
print(df["clean_text"].iloc[0])
```

```python
df.to_csv("preprocessed_fake_news.csv", index=False)

print("Đã lưu dữ liệu tiền xử lý!")
```
